```python
dupe_ids = combined_df[combined_df.duplicated(subset=['CustomerId'], keep=False)].sort_values('CustomerId')
dupe_ids
```

```python
#Merginng the jones pair 
group_d_id = 15567632
d_rows =combined_df[combined_df['CustomerId'] == group_d_id]
row0 = d_rows.iloc[[0]].reset_index(drop=True)
row1 = d_rows.iloc[[1]].reset_index(drop=True)
merged_row = row0.combine_first(row1)

combined_df = combined_df[combined_df['CustomerId'] != group_d_id]
combined_df = pd.concat([combined_df, merged_row], ignore_index=True)
print('After group D', len(combined_df))
```

```python
#Mokoena roes: keep the complete row 
complete_id = 15566621
complete_rows = combined_df[combined_df['CustomerId'] == complete_id].copy()
complete_rows['missing_count'] = complete_rows.isnull().sum(axis=1)
complete_kept = complete_rows.sort_values('missing_count').drop_duplicates(subset = 'CustomerId', keep='first')
complete_drop_index = complete_rows.index.difference(complete_kept.index)
combined_df = combined_df.drop(index=complete_drop_index)
```

```python
print('After keeping the complete rows and droping the incomplet', len(combined_df))
```

```python
#group b, the othwe 5, flag unreliable, drop dunlicates only within the subset
unreliable_ids = [15570651, 15591784, 15627259, 15636646, 15660454]
combined_df['unreliable_rows'] = combined_df['CustomerId'].isin(unreliable_ids)

unreliable_rows = combined_df[combined_df['CustomerId'].isin(unreliable_ids)]
kept = unreliable_rows.drop_duplicates(subset='CustomerId', keep='first')
rows_to_drop = unreliable_rows.index.difference(kept.index)
combined_df = combined_df.drop(index=rows_to_drop)
print('After unreliable group', len(combined_df))
 
```

```python
#group c flag only, dont drop anything

collission_ids = [15612193, 15619704]
combined_df['id_collision_flag'] = combined_df['CustomerId'].isin(collission_ids)
print('final_rows', len(combined_df))
```
